# Transformers

- Transformers are the foundation of many modern NLP models, including BERT, GPT, and T5.

- Before Transformers, we used models such as RNNs and LSTMs for NLP tasks.

## 1. Load a Transformer

We use **DistilBERT**, a  Transformer fine-tuned for positive/negative sentiment classification.

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_name = "distilbert-base-uncased-finetuned-sst-2-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSequenceClassification.from_pretrained(model_name)
model.eval()

# Look inside the Transformer architecture.
print("Transformer layers:", model.config.n_layers)
print("Attention heads per layer:", model.config.n_heads)
print("Hidden size:", model.config.dim)
```

## 2. Tokenizing the text

```python
sentence = "The movie was not good."

# Tokenization: converts text into token IDs
inputs = tokenizer(sentence, return_tensors="pt")

# [CLS] = start token, [SEP] = end token
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

print("Tokens:", tokens)
```

## 3. Embeddings and self-attention

Each token ID becomes a **vector (embedding)**. Inside each Transformer layer, **self-attention** lets a token use information from other tokens in the same sentence.

```python
import torch

# Enable attention weights
model.set_attn_implementation("eager")

with torch.no_grad():
    embeddings = model.distilbert.embeddings(inputs["input_ids"])
    outputs = model(**inputs, output_attentions=True)

# Attention from the first layer and first head
attention = outputs.attentions[0][0, 0]

print("Embeddings shape:", embeddings.shape)
print("Attention shape:", attention.shape)
print("Tokens:", tokens)
print("Attention weights:\n", attention)
```

## 4. Classification: From Transformer to prediction

The Transformer produces contextual representations. The **classification head** turns them into two scores (*logits*). **Softmax** converts these into probabilities that sum to 1.

```python
# Get prediction scores from the Transformer
with torch.no_grad():
    logits = model(**inputs).logits

# Convert scores into probabilities
probs = torch.softmax(logits, dim=1)[0]

# Display the prediction
prediction = probs.argmax().item()

print("Prediction:", model.config.id2label[prediction])
print("Confidence:", round(probs[prediction].item(), 3))
```


### Task: Exploring Transformer Layers and Attention Heads

Load the pretrained **`bert-base-uncased`** model using Hugging Face Transformers.

2. Calculate and print the total number of attention heads across all Transformer layers.
3. Using the sentence `"Transformers use self-attention to understand language."`, extract the attention weights from the **third Transformer layer and second attention head**.

## Task solution
Run the installation cell first. The solution below uses its own BERT variables, so it can also be run without the DistilBERT demonstration. An internet connection is needed on the first run to download the pretrained model and tokenizer.

```python
from transformers import AutoTokenizer, AutoModel
import torch

# Load pretrained BERT. Eager attention makes the attention weights available.
bert_name = "bert-base-uncased"
bert_tokenizer = AutoTokenizer.from_pretrained(bert_name)
bert_model = AutoModel.from_pretrained(bert_name, attn_implementation="eager")
bert_model.eval()  # Disable dropout for reproducible inference.

# 1. Read the architecture from BERT's configuration.
num_layers = bert_model.config.num_hidden_layers
heads_per_layer = bert_model.config.num_attention_heads
print("Transformer layers:", num_layers)
print("Attention heads per layer:", heads_per_layer)

# 2. Count heads across all Transformer layers.
total_heads = num_layers * heads_per_layer
print("Total attention heads:", total_heads)

# 3. Extract the third layer, second head for the first (only) sentence.
bert_sentence = "Transformers use self-attention to understand language."
bert_inputs = bert_tokenizer(bert_sentence, return_tensors="pt")
bert_tokens = bert_tokenizer.convert_ids_to_tokens(bert_inputs["input_ids"][0].tolist())

with torch.no_grad():
    bert_outputs = bert_model(**bert_inputs, output_attentions=True, return_dict=True)

# Each layer has shape: (batch_size, num_heads, sequence_length, sequence_length).
# Python uses zero-based indexing: layer 3 -> 2, sentence 1 -> 0, head 2 -> 1.
bert_attention = bert_outputs.attentions[2][0, 1].cpu()

print("Tokens:", bert_tokens)
print("Attention matrix shape:", tuple(bert_attention.shape))
print("Attention weights (layer 3, head 2):\n", bert_attention)

# Check the matrix dimensions and attention normalization.
assert len(bert_outputs.attentions) == num_layers
assert bert_attention.shape == (len(bert_tokens), len(bert_tokens))
assert torch.allclose(
    bert_attention.sum(dim=-1), torch.ones(len(bert_tokens)), atol=1e-5
)
```

### 4. What do the rows and columns represent?

- **Rows (query tokens):** the token whose representation is being updated.
- **Columns (key tokens):** the tokens that the row token attends to.
- **Entry `bert_attention[i, j]`:** how much attention token `i` gives token `j` in the third layer's second head. A larger value means a larger weight for token `j` when combining value vectors for token `i` in this head.

Both axes follow the printed token order, including `[CLS]`, `[SEP]`, punctuation, and any WordPiece subwords. The matrix is square because every token can attend to all tokens in the same sentence. Each row sums to approximately 1 with dropout disabled; the matrix need not be symmetric.

**Expected architecture output:** 12 Transformer layers, 12 attention heads per layer, and **12 × 12 = 144 total attention heads**. The attention values themselves are computed when you run the code.

```python
# Optional: visualize the extracted matrix.
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 8))
image = ax.imshow(bert_attention.numpy(), cmap="viridis", vmin=0)
ax.set_xticks(range(len(bert_tokens)))
ax.set_xticklabels(bert_tokens, rotation=60, ha="right")
ax.set_yticks(range(len(bert_tokens)))
ax.set_yticklabels(bert_tokens)
ax.set_xlabel("Key tokens (tokens being attended to)")
ax.set_ylabel("Query tokens (tokens giving attention)")
ax.set_title("BERT: third Transformer layer, second attention head")
fig.colorbar(image, ax=ax, label="Attention weight")
fig.tight_layout()
plt.show()
```
