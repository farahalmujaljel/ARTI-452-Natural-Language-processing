Task 01

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("seriousran/appletwittersentimenttexts")

print("Path to dataset files:", path)
```

```python
import os
print(os.listdir(path))
```

```python
import pandas as pd
import re
import os
from collections import Counter


files = os.listdir(path)
csv_file = [file for file in files if file.endswith(".csv")][0]


df = pd.read_csv(os.path.join(path, csv_file))


print(df.columns)
print(df.head())
```

```python
text_column = [col for col in df.columns if "text" in col.lower()][0]

tweets = df[text_column].dropna().astype(str)

hashtags = []

for tweet in tweets:
    hashtags.extend(re.findall(r"#\w+", tweet.lower()))

print("Total number of hashtags:", len(hashtags))

top_10 = Counter(hashtags).most_common(10)

print("\nTop 10 hashtags:")
for hashtag, count in top_10:
    print(hashtag, ":", count)
```

```python
import re

text = "This year is 2021"

pattern = re.compile(r"\d+")

print(type(pattern))

updated_text = pattern.sub("2022", text)

print(updated_text)
```
